```python

'''
EdgeTx/OpenTx radio serial port is known to have vid == 0x0483, pid = 0x5740
'''
from serial.tools.list_ports import comports
portList = list(comports())
print(portList)
radioportList = []
for port in portList:
    print(port.vid, port.pid, port.manufacturer, port.device)
    if port.vid == 0x0483 and port.pid == 0x5740:
            if port.manufacturer in ('EdgeTX', 'OpenTX'):
                radioportList.append(port.device)
            radioportList.append(port.device)

print("Radio ports found:", radioportList)
```
